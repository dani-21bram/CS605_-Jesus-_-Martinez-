```python
class ContactNode:
  #here we just define a class for each individual(contact)
    def __init__(self, contact_name, phone_number):

        self.contact_name = contact_name
        self.phone_number = phone_number
        self.before_this_contact = None   # here the contacts that come before alphabetically (left)
        self.after_this_contact = None    #  here the contacts that come after alphabetically (right)

#here is the defenitoon with some of target parameter, those bieng contac name and theri phone number
class PhonebookTree:
    def __init__(self):
      #emtoty one to then keeo adding more
        self.first_contact = None  # first contact to be created

    def add_contact(self, contact_name, phone_number):
        new_contact = ContactNode(contact_name, phone_number)

        # here below we just check if phonebook is empty, then it would become our first contact
        if self.first_contact is None:
            self.first_contact = new_contact
            print(f"Added {contact_name} as first contact")
        else:
          #finding the corect position i the tree
            self._find_spot_for_new_contact(self.first_contact, new_contact)

    def _find_spot_for_new_contact(self, current_contact, new_contact):
       #see where to write the new contact number
        # here we compare names to see where the new contact belongs
        if new_contact.contact_name < current_contact.contact_name:
            # updating the new names in a way that itcomes before the current
            # contact alphabetically
            if current_contact.before_this_contact is None:
                # if found then its inserted here
                current_contact.before_this_contact = new_contact
                print(f"Added {new_contact.contact_name} - comes before {current_contact.contact_name}")
            else:
                # we just keep looking in the past section if not found
                self._find_spot_for_new_contact(current_contact.before_this_contact, new_contact)
        elif new_contact.contact_name > current_contact.contact_name:
            # New name comes after current contact alphabetically
            if current_contact.after_this_contact is None:
                current_contact.after_this_contact = new_contact
                print(f"Added {new_contact.contact_name} - comes after {current_contact.contact_name}")
            else:
                self._find_spot_for_new_contact(current_contact.after_this_contact, new_contact)
        else:
            current_contact.phone_number = new_contact.phone_number
            print(f"Updated {contact_name}'s phone number")
#same logic just the opposite ^



    def find_phone_number(self, name_to_find):
        return self._search_for_contact(self.first_contact, name_to_find)
    #searcehs for contatcs(function)
    def _search_for_contact(self, current_contact, name_to_find):

        # In case  we've reached the end/empty
        if current_contact is None:
            return None

        # Founds the person that is being searched
        if name_to_find == current_contact.contact_name:
            return current_contact.phone_number
        # Name comes before current contact - look in the 'before' section
        elif name_to_find < current_contact.contact_name:
            return self._search_for_contact(current_contact.before_this_contact, name_to_find)
        # Name comes after current contact - look in the 'after' section
        else:
            return self._search_for_contact(current_contact.after_this_contact, name_to_find)

    def show_all_contacts_sorted(self):

        #Display all contacts in perfect alphabetical order.

        print("\nPhonebook content (Alphabetical Order):")
        self._read_contacts_in_order(self.first_contact)

    def _read_contacts_in_order(self, contact):

        if contact is not None:
            # First, show all contacts that come before this one
            self._read_contacts_in_order(contact.before_this_contact)

            # Then show this contact
            print(f"{contact.contact_name}: {contact.phone_number}")

            # Finally, show all contacts that come after this one
            self._read_contacts_in_order(contact.after_this_contact)

    # BONUS: Remove a contact
    def remove_contact(self, name_to_remove):

        self.first_contact = self._remove_contact_helper(self.first_contact, name_to_remove)

    def _remove_contact_helper(self, current_contact, name_to_remove):
        # Contact not found
        if current_contact is None:
            print(f"{name_to_remove} not found in phonebook")
            return None

        # Still looking for the contact to remove
        if name_to_remove < current_contact.contact_name:
            current_contact.before_this_contact = self._remove_contact_helper(current_contact.before_this_contact, name_to_remove)
        elif name_to_remove > current_contact.contact_name:
            current_contact.after_this_contact = self._remove_contact_helper(current_contact.after_this_contact, name_to_remove)
        else:
            # contact founed then has to be removed
            print(f"Removing {name_to_remove} from phonebook")

            # Case 1: Contact has no connections (leaf node)
            if current_contact.before_this_contact is None and current_contact.after_this_contact is None:
                return None

            # Case 2: Contact has only one connection
            elif current_contact.before_this_contact is None:
                return current_contact.after_this_contact
            elif current_contact.after_this_contact is None:
                return current_contact.before_this_contact

            # Case 3: Contact has two connections - need to find a replacement
            else:
                # Find the next contact in alphabetical order
                next_in_line = self._find_next_contact(current_contact.after_this_contact)
                # Move that contact here
                current_contact.contact_name = next_in_line.contact_name
                current_contact.phone_number = next_in_line.phone_number
                # Remove the duplicates
                current_contact.after_this_contact = self._remove_contact_helper(current_contact.after_this_contact, next_in_line.contact_name)

        return current_contact

    def _find_next_contact(self, contact):
        current = contact
        while current.before_this_contact is not None:
            current = current.before_this_contact
        return current


# testing the phonebook
def demo_phonebook():

    # Creating  a new phonebook
    my_phonebook = PhonebookTree()

    print("\n adding contacts")
    # Adding contacts
    friends = [
        ("Dani", "555-7890"),
        ("Alma", "555-1234"),
        ("Bryan", "555-5678"),
        ("Juan", "555-2468"),
        ("Emely", "555-1357")
    ]

    for name, phone in friends:
        my_phonebook.add_contact(name, phone)

    print("\n looking up by numbers")
    # finding numbers
    people_to_call = ["Dani", "Alma", "Bryan", "Juan"]

    for person in people_to_call:
        number = my_phonebook.find_phone_number(person)
        if number:
            print(f"Found {person}! Call them at: {number}")
        else:
            print(f"Sorry, {person} isn't in my phonebook")

    print("\n observing the entire phonebook")
    # Show everything in order
    my_phonebook.show_all_contacts_sorted()

    print("\n extra(bonus) removing someone ")
    # Test removing someone
    print("Before removal:")
    my_phonebook.show_all_contacts_sorted()

    my_phonebook.remove_contact("Juan")
    print("After removing Juan:")
    my_phonebook.show_all_contacts_sorted()


if __name__ == "__main__":
    demo_phonebook()
```
